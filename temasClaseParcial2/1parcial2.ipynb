```python
import cv2
import numpy as np
import os
import pandas as pd
from skimage.feature import graycomatrix, graycoprops
from skimage import img_as_ubyte

# Carpeta donde están tus imágenes

image = cv2.imread('cereal.png', cv2.IMREAD_GRAYSCALE)

# 1. Media
mean = np.mean(image)

# 2. Varianza
variance = np.var(image)

# 3. Desviación estándar
std_dev = np.std(image)

# 4. Entropía
# Convertir la imagen a tipo float para evitar problemas con valores negativos
image_float = np.float32(image) + 1e-5  # Añadir un pequeño valor para evitar log(0)
entropy = -np.sum(image_float * np.log(image_float))

print(f"Media: {mean}")
print(f"Varianza: {variance}")
print(f"Desviación Estándar: {std_dev}")
print(f"Entropía: {entropy}")

import numpy as np
import cv2
from skimage.feature import graycomatrix
from skimage.feature import graycoprops
from skimage import img_as_ubyte

'''
a partir de la versión 0.19 las funciones tienen nombre de: graycomatrixy graycoprops
versión 1.0 las funciones tienen nombre de: greycomatrix, greycoprops
'''

# Cargar imagen en escala de grises


# Convertir la imagen a un tipo adecuado para GLCM
image = img_as_ubyte(image)

# Calcular la matriz de co-ocurrencia con un desplazamiento de 1 píxel en dirección horizontal
glcm = graycomatrix(image, distances=[1], angles=[0], symmetric=True, normed=True)

# 1. Contraste
contrast = graycoprops(glcm, prop='contrast')

# 2. Homogeneidad
homogeneity = graycoprops(glcm, prop='homogeneity')

# 3. Disimilitud
dissimilarity = graycoprops(glcm, prop='dissimilarity')

# 4. Energía
energy = graycoprops(glcm, prop='energy')

# 5. Correlación
correlation = graycoprops(glcm, prop='correlation')

# 6. Media de la GLCM
mean_glcm = np.mean(glcm)

# 7. Desviación estándar de la GLCM
std_dev_glcm = np.std(glcm)

# 8. Entropía de la GLCM
# Calcular entropía de la GLCM (una aproximación simple)
glcm_flat = glcm.flatten()
glcm_flat = glcm_flat[glcm_flat > 0]  # Eliminar ceros para evitar log(0)
entropy_glcm = -np.sum(glcm_flat * np.log(glcm_flat))

# Mostrar resultados
print(f"Contraste: {contrast}")
print(f"Homogeneidad: {homogeneity}")
print(f"Disimilitud: {dissimilarity}")
print(f"Energía: {energy}")
print(f"Correlación: {correlation}")
print(f"Media de la GLCM: {mean_glcm}")
print(f"Desviación Estándar de la GLCM: {std_dev_glcm}")
print(f"Entropía de la GLCM: {entropy_glcm}")

import cv2
import numpy as np
import matplotlib.pyplot as plt

# Cargar la imagen




imagen_suavizada = cv2.GaussianBlur(image, (5, 5), 0)

# Umbralizar la imagen para obtener una imagen binaria (en caso de que sea una imagen de fondo blanco y negro)
_, binary_image = cv2.threshold(imagen_suavizada, 107, 255, cv2.THRESH_BINARY)
imag = cv2.Canny(binary_image, 100, 200)

# Calcular los momentos de la imagen binaria
moments = cv2.moments(imag)

# Calcular los Momentos de Hu
hu_moments_1 = cv2.HuMoments(moments)
print(hu_moments_1)

# Mostrar los Momentos de Hu
print("Momentos de Hu:")
for i in range(7):
    print(f'Momento {i+1}: {hu_moments_1[i][0]}')

# Puedes visualizar la imagen original
# cv2.imshow("Imagen", image)
# cv2.waitKey(0)
# cv2.destroyAllWindows()

plt.title('Imagen')
plt.imshow(imag, cmap='gray')
plt.axis('off')  # Ocultar los ejes
plt.show()



import cv2
import numpy as np
import matplotlib.pyplot as plt

# Cargar la imagen


# Umbralizar la imagen para obtener una imagen binaria (en caso de que sea una imagen de fondo blanco y negro)
_, binary_image = cv2.threshold(image, 127, 255, cv2.THRESH_BINARY)

# Calcular los momentos de la imagen binaria
moments = cv2.moments(binary_image)

# Calcular los Momentos de Hu
hu_moments_2 = cv2.HuMoments(moments)
print(hu_moments_2)

# Mostrar los Momentos de Hu
print("Momentos de Hu:")
for i in range(7):
    print(f'Momento {i+1}: {hu_moments_2[i][0]}')

# Puedes visualizar la imagen original
# cv2.imshow("Imagen", image)
# cv2.waitKey(0)
# cv2.destroyAllWindows()

plt.title('Imagen')
plt.imshow(image, cmap='gray')
plt.axis('off')  # Ocultar los ejes
plt.show()



distance_1 = np.linalg.norm(hu_moments_1 - hu_moments_2)

print(f"Distancia Euclidiana entre los momentos de Hu: {distance_1}")



import cv2
import numpy as np
import matplotlib.pyplot as plt

# Cargar la imagen

# Umbralizar la imagen para obtener una imagen binaria (en caso de que sea una imagen de fondo blanco y negro)
_, binary_image = cv2.threshold(image, 127, 255, cv2.THRESH_BINARY)

# Calcular los momentos de la imagen binaria
moments = cv2.moments(binary_image)

# Calcular los Momentos de Hu
hu_moments_3 = cv2.HuMoments(moments)
print(hu_moments_3)

# Mostrar los Momentos de Hu
print("Momentos de Hu:")
for i in range(7):
    print(f'Momento {i+1}: {hu_moments_3[i][0]}')

# Puedes visualizar la imagen original
# cv2.imshow("Imagen", image)
# cv2.waitKey(0)
# cv2.destroyAllWindows()

plt.title('Imagen')
plt.imshow(image, cmap='gray')
plt.axis('off')  # Ocultar los ejes
plt.show()

print(hu_moments_1)
print()
print(hu_moments_2)
print()
print(hu_moments_3)


distance_2 = np.linalg.norm(hu_moments_2 - hu_moments_3)

print(f"Distancia Euclidiana entre los momentos de Hu: {distance_2}")


print(distance_1)
print(distance_2)


import cv2
import numpy as np
import os


# Cargar imagen

gray = image  # 'image' is already grayscale

# Crear el descriptor HOG
hog = cv2.HOGDescriptor()

# Calcular el descriptor HOG
hog_features = hog.compute(gray)

# Mostrar los resultados

print("Características HOG extraídas: Este método devuelve un conjunto de características que representan la distribución de gradientes en diferentes direcciones en la imagen.\n", hog_features)


import numpy as np
import matplotlib.pyplot as plt

# Usar la imagen en escala de grises ya cargada
# gray ya contiene la imagen en escala de grises

# Crear el detector KAZE
kaze = cv2.KAZE_create()

# Detectar puntos clave y calcular descriptores
keypoints, descriptors = kaze.detectAndCompute(gray, None)

#keypoints: Es una lista de los puntos clave detectados en la imagen.
#Los puntos clave son ubicaciones en la imagen donde se pueden observar características interesantes,
#como bordes, esquinas o áreas con texturas únicas.
#descriptors: Son vectores numéricos que representan las características locales alrededor de cada punto clave.
#Estos descriptores son útiles para comparar puntos clave entre diferentes imágenes.

# Dibujar los puntos clave sobre la imagen original
img_keypoints = cv2.drawKeypoints(image, keypoints, None, color=(0, 255, 0))

#None: Este argumento se refiere a la imagen de salida (si se pasa None, OpenCV crea una nueva imagen para mostrar los puntos clave).
#color=(0, 255, 0): Este es el color que se usará para dibujar los puntos clave. En este caso, es verde (el valor en BGR de verde es (0, 255, 0)).

# Mostrar la imagen con los puntos clave
# cv2.imshow('KAZE Keypoints', img_keypoints)
# cv2.waitKey(0)
# cv2.destroyAllWindows()

# Mostrar la imagen original
plt.imshow(image)
plt.title("Original")
plt.axis('off')
plt.show()

# Mostrar la imagen con los puntos clave
plt.imshow(img_keypoints)
plt.title("KAZE Keypoints")
plt.axis('off')
plt.show()

print('Descriptores de la imagen',descriptors)


import cv2
import numpy as np
import matplotlib.pyplot as plt

# Cargar la imagen (asegúrate de tener la imagen en la misma carpeta o proporcionar la ruta completa)

# Definir los valores de sigma
sigma_values = [0.5, 2, 5.5, 8, 11, 15.5, 20, 24.5]

# Crear una figura para mostrar las imágenes filtradas
plt.figure(figsize=(12, 4))

# Aplicar el filtro gaussiano con los diferentes valores de sigma
for i, sigma in enumerate(sigma_values):
    # Calcular el tamaño del kernel en función de la desviación estándar
    ksize = int(2 * (sigma * 3) + 1)  # Se suele usar un tamaño de kernel de 6 veces sigma más 1
    # Asegurar que ksize es impar
    ksize = ksize + 1 if ksize % 2 == 0 else ksize # If ksize is even, add 1 to make it odd
    # Aplicar el filtro gaussiano
    filtered_image = cv2.GaussianBlur(image, (ksize, ksize), sigma)

    # Guardar la imagen filtrada con un nombre específico
    output_filename = f'filtered_image_sigma_{sigma}.jpg'
    cv2.imwrite(output_filename, filtered_image)

    # Mostrar la imagen filtrada
    plt.subplot(1, len(sigma_values), i + 1)
    plt.imshow(filtered_image, cmap='gray')
    plt.title(f'FG (σ={sigma})')
    plt.axis('off')

# Mostrar la gráfica
plt.show()

print("Las imágenes filtradas se han guardado correctamente.")


import cv2
import numpy as np
import matplotlib.pyplot as plt

# Cargar la imagen (asegúrate de tener la imagen en la misma carpeta o proporcionar la ruta completa)


# Definir los valores de sigma
sigma_values = [0.5, 2, 5.5, 8, 11, 15.5, 20, 24.5]

# Crear un diccionario para almacenar las imágenes filtradas
filtered_images = {}

# Crear una figura para mostrar las imágenes filtradas
plt.figure(figsize=(12, 4))

# Aplicar el filtro gaussiano con los diferentes valores de sigma
for i, sigma in enumerate(sigma_values):
    # Calcular el tamaño del kernel en función de la desviación estándar
    ksize = int(2 * (sigma * 3) + 1)  # Se suele usar un tamaño de kernel de 6 veces sigma más 1
    # Asegurar que ksize es impar
    ksize = ksize + 1 if ksize % 2 == 0 else ksize # If ksize is even, add 1 to make it odd
    # Aplicar el filtro gaussiano
    filtered_image = cv2.GaussianBlur(image, (ksize, ksize), sigma)

    # Guardar la imagen filtrada
    filtered_images[sigma] = filtered_image

    # Mostrar la imagen filtrada
    plt.subplot(1, len(sigma_values), i + 1)
    plt.imshow(filtered_image, cmap='gray')
    plt.title(f'Filtro Gaussiano (σ={sigma})')
    plt.axis('off')

# Mostrar la gráfica
plt.show()

# Calcular y mostrar las diferencias entre las imágenes
for i, sigma1 in enumerate(sigma_values):
    for j, sigma2 in enumerate(sigma_values):
        if sigma1 < sigma2:  # Evitar comparar una imagen consigo misma
            diff = cv2.absdiff(filtered_images[sigma1], filtered_images[sigma2])  # Diferencia absoluta entre imágenes
            diff_filename = f'difference_sigma_{sigma1}_vs_{sigma2}.jpg'
            cv2.imwrite(diff_filename, diff)  # Guardar la imagen de la diferencia

            # Mostrar la diferencia
            plt.figure(figsize=(6, 6))
            plt.imshow(diff, cmap='gray')
            plt.title(f'Diferencia entre σ={sigma1} y σ={sigma2}')
            plt.axis('off')
            plt.show()

print("Las diferencias entre las imágenes han sido calculadas y guardadas correctamente.")

import cv2
import numpy as np
import matplotlib.pyplot as plt

# Cargar la imagen en escala de grises

# Crear el detector de características SIFT
sift = cv2.SIFT_create()

# Detectar los puntos clave (keypoints) y calcular descriptores
keypoints, descriptors = sift.detectAndCompute(image, None)

# Dibujar los puntos clave en la imagen original
image_with_keypoints = cv2.drawKeypoints(image, keypoints, None)

# Mostrar la imagen con los puntos clave detectados
plt.imshow(image_with_keypoints)
plt.title("Puntos clave detectados con SIFT")
plt.axis('off')  # Ocultar los ejes
plt.show()
import cv2
import numpy as np

# Cargar las dos imágenes
img1 = cv2.imread('cereal.png', cv2.IMREAD_GRAYSCALE)  # Imagen de consulta
img2 = cv2.imread('barritas.png', cv2.IMREAD_GRAYSCALE)  # Imagen de referencia

# Crear el detector SIFT
sift = cv2.SIFT_create()

# Detectar puntos clave y descriptores
keypoints1, descriptors1 = sift.detectAndCompute(img1, None)
keypoints2, descriptors2 = sift.detectAndCompute(img2, None)

# Crear el objeto de coincidencias con KNN
bf = cv2.BFMatcher()

# Coincidencias más cercanas con KNN
matches = bf.knnMatch(descriptors1, descriptors2, k=2)

# Aplicar la prueba de ratio de Lowe para filtrar coincidencias
good_matches = []
for m, n in matches:
    if m.distance < 0.75 * n.distance:
        good_matches.append(m)

# Dibujar las coincidencias
img_matches = cv2.drawMatches(img1, keypoints1, img2, keypoints2, good_matches, None, flags=cv2.DrawMatchesFlags_NOT_DRAW_SINGLE_POINTS)

# Mostrar las coincidencias
# cv2.imshow("Coincidencias", img_matches)
# cv2.waitKey(0)
# cv2.destroyAllWindows()

plt.imshow(img_matches)
plt.title("Coincidencias")
plt.axis('off')  # Ocultar los ejes
plt.show()


sift = cv2.SIFT_create(nfeatures=500)  # Limitar a 500 características






import cv2
import numpy as np
import matplotlib.pyplot as plt

# Crear el objeto ORB
orb = cv2.ORB_create()

# Detectar los puntos clave y calcular los descriptores
keypoints, descriptors = orb.detectAndCompute(image, None)

# Dibujar los puntos clave sobre la imagen
image_with_keypoints = cv2.drawKeypoints(image, keypoints, image, (255, 0, 0), flags=cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS)
# None imagen de salida, donde se dibujarán los puntos clave. No se esta guardando.
#Si se quiere usar una imagen nueva para los puntos.

#cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS es una opción que dibuja los puntos clave con una representación más detallada,
#mostrando no solo los puntos clave (normalmente representados como círculos), sino también la orientación
#y la escala de los puntos clave. También verá una línea que indica la orientación, mejora la visualizado .
#si se usa cv2.DRAW_MATCHES_FLAGS_DEFAULT, dibujará solo un círculo simple en la posición de cada punto clave
#sin mostrar la orientación ni la escala.

# Mostrar la imagen con los puntos clave
plt.imshow(image_with_keypoints)
plt.title("Puntos clave detectados con ORB")
plt.axis('off')
plt.show()

# Si deseas guardar la imagen con los puntos clave
cv2.imwrite('image_with_orb_keypoints.jpg', image_with_keypoints)



import cv2
import numpy as np

# Cargar dos imágenes consecutivas de un video
frame1 = cv2.imread('barritas.png', cv2.IMREAD_GRAYSCALE)
frame2 = cv2.imread('cereal.png', cv2.IMREAD_GRAYSCALE)

# Redimensionar frame2 para que coincida con frame1
if frame1.shape != frame2.shape:
	frame2 = cv2.resize(frame2, (frame1.shape[1], frame1.shape[0]))

# Calcular el flujo óptico usando el método de Lucas-Kanade
flow = cv2.calcOpticalFlowFarneback(frame1, frame2, None, 0.5, 3, 15, 3, 5, 1.2, 0)

# Convertir el flujo óptico a un formato visualizable
mag, ang = cv2.cartToPolar(flow[..., 0], flow[..., 1])
hsv = np.zeros_like(frame1, dtype=np.float32)
# Apilar imágenes en escala de grises para tener 3 canales
hsv = np.stack((hsv,hsv,hsv), axis=-1)
hsv[..., 1] = 255
hsv[..., 0] = ang * 180 / np.pi / 2
hsv[..., 2] = cv2.normalize(mag, None, 0, 255, cv2.NORM_MINMAX)
flow_rgb = cv2.cvtColor(hsv, cv2.COLOR_HSV2BGR)

# Mostrar el flujo óptico
# cv2.imshow('Optical Flow', flow_rgb)
# cv2.waitKey(0)
# cv2.destroyAllWindows()

plt.title('Optical Flow')
plt.imshow(flow_rgb, cmap=None)
plt.axis('off')

plt.show()


import cv2
import numpy as np
import pandas as pd
import os

# Lista de nombres de imágenes
imagenes = ['cereal.png', 'barritas.png', 'doritos.png', 'embutidos.png', 'galletas.png',
            'gaseosa.png', 'papas.png', 'papitas.png', 'pizza.png', 'varias.png']

# Crear DataFrame vacío
resultados = pd.DataFrame(columns=[
    'Imagen', 'SIFT_KP', 'ORB_KP', 'Coincidencias', 'Distancia_Promedio', 'Flujo_Magnitud', 'Flujo_Dirección'
])
def analizar_imagen(imagen_path, referencia_path=None):
    img = cv2.imread(imagen_path, cv2.IMREAD_GRAYSCALE)
    sift = cv2.SIFT_create()
    orb = cv2.ORB_create()

    # SIFT
    kp_sift, desc_sift = sift.detectAndCompute(img, None)
    sift_kp = len(kp_sift)

    # ORB
    kp_orb, desc_orb = orb.detectAndCompute(img, None)
    orb_kp = len(kp_orb)

    # Coincidencias con referencia
    coincidencias = 0
    distancia_promedio = 0
    if referencia_path:
        ref = cv2.imread(referencia_path, cv2.IMREAD_GRAYSCALE)
        kp_ref, desc_ref = sift.detectAndCompute(ref, None)
        bf = cv2.BFMatcher()
        matches = bf.knnMatch(desc_sift, desc_ref, k=2)
        good = [m for m, n in matches if m.distance < 0.75 * n.distance]
        coincidencias = len(good)
        if good:
            distancia_promedio = np.mean([m.distance for m in good])

    # Optical Flow (comparar con referencia si existe)
    flujo_mag = 0
    flujo_ang = 0
    if referencia_path:
        ref = cv2.imread(referencia_path, cv2.IMREAD_GRAYSCALE)
        if img.shape != ref.shape:
            ref = cv2.resize(ref, (img.shape[1], img.shape[0]))
        flow = cv2.calcOpticalFlowFarneback(ref, img, None, 0.5, 3, 15, 3, 5, 1.2, 0)
        mag, ang = cv2.cartToPolar(flow[..., 0], flow[..., 1])
        flujo_mag = np.mean(mag)
        flujo_ang = np.mean(ang)

    return sift_kp, orb_kp, coincidencias, distancia_promedio, flujo_mag, flujo_ang
for img in imagenes:
    referencia = 'cereal.png' if img != 'cereal.png' else None  # Usar cereal como referencia
    sift_kp, orb_kp, match_count, match_dist, flow_mag, flow_ang = analizar_imagen(img, referencia)
    resultados.loc[len(resultados)] = [img, sift_kp, orb_kp, match_count, match_dist, flow_mag, flow_ang]
    print(resultados)
# resultados.to_csv('metricas_comparativas.csv', index=False)
```
