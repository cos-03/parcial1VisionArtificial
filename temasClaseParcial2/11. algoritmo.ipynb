```python
import cv2
import numpy as np
import os


# Cargar imagen
image_path = 'barritas.png'  # Cambia esta ruta si tu imagen está en otra ubicación
image = cv2.imread(image_path)
if image is None:
	raise FileNotFoundError(f"No se pudo cargar la imagen. Verifica la ruta: '{image_path}'")
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Crear el descriptor HOG
hog = cv2.HOGDescriptor()

# Calcular el descriptor HOG
hog_features = hog.compute(gray)

# Mostrar los resultados

print("Características HOG extraídas: Este método devuelve un conjunto de características que representan la distribución de gradientes en diferentes direcciones en la imagen.\n", hog_features)


import numpy as np
import matplotlib.pyplot as plt

# Usar la imagen en escala de grises ya cargada
# gray ya contiene la imagen en escala de grises

# Crear el detector KAZE
kaze = cv2.KAZE_create()

# Detectar puntos clave y calcular descriptores
keypoints, descriptors = kaze.detectAndCompute(gray, None)

#keypoints: Es una lista de los puntos clave detectados en la imagen.
#Los puntos clave son ubicaciones en la imagen donde se pueden observar características interesantes,
#como bordes, esquinas o áreas con texturas únicas.
#descriptors: Son vectores numéricos que representan las características locales alrededor de cada punto clave.
#Estos descriptores son útiles para comparar puntos clave entre diferentes imágenes.

# Dibujar los puntos clave sobre la imagen original
img_keypoints = cv2.drawKeypoints(image, keypoints, None, color=(0, 255, 0))

#None: Este argumento se refiere a la imagen de salida (si se pasa None, OpenCV crea una nueva imagen para mostrar los puntos clave).
#color=(0, 255, 0): Este es el color que se usará para dibujar los puntos clave. En este caso, es verde (el valor en BGR de verde es (0, 255, 0)).

# Mostrar la imagen con los puntos clave
# cv2.imshow('KAZE Keypoints', img_keypoints)
# cv2.waitKey(0)
# cv2.destroyAllWindows()

# Mostrar la imagen original
plt.imshow(image)
plt.title("Original")
plt.axis('off')
plt.show()

# Mostrar la imagen con los puntos clave
plt.imshow(img_keypoints)
plt.title("KAZE Keypoints")
plt.axis('off')
plt.show()

print('Descriptores de la imagen',descriptors)
```

Explicación:
Cargar la imagen: Primero, cargamos la imagen en escala de grises, ya que los algoritmos KAZE y AKAZE funcionan mejor en este tipo de imágenes.

Detector KAZE: Creamos un objeto kaze utilizando cv2.KAZE_create() y usamos kaze.detectAndCompute() para detectar los puntos clave y calcular los descriptores.

puntos clave: Usamos cv2.drawKeypoints() para dibujar los puntos clave detectados sobre la imagen.
