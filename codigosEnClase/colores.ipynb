Este notebook muestra una guía visual de diferentes cmap disponibles en matplotlib,
aplicados a una misma matriz aleatoria.
keyboard_arrow_down Catálogo de Colormaps en Matplotlib
¿Qué es cmap?
cmap significa colormap (mapa de colores). Es un parámetro de Matplotlib que define
cómo se asignan colores a valores numéricos en una visualización (imágenes, mapas de
calor, gráficos de dispersión, etc.).
Imagina que tienes una matriz con valores entre 0 y 1:
Si usas cmap="gray", los valores bajos se pintan en negro y los altos en blanco.
Si usas cmap="hot", los valores bajos se pintan en negro, medios en rojo-amarillo, y altos en
blanco.
Si usas cmap="viridis", los valores van de azul → verde → amarillo.
En otras palabras:
👉 cmap es la regla de colores que traduce números en colores.
Tipos de cmap en Matplotlib
Secuenciales → Para valores que solo crecen o decrecen. Ej: "viridis", "plasma", "Blues",
"Greens"
Divergentes → Para valores que tienen un punto medio importante. Ej: "coolwarm", "bwr",
"seismic"
Categóricos (cualitativos) → Para datos de categorías distintas. Ej: "tab10", "Set1", "Pastel1"
En resumen:
cmap controla la paleta de colores usada para representar números.
Sirve para mapas de calor, imágenes, scatter plots y superficies 3D.
Cambiar cmap cambia totalmente la percepción de los datos.


```python

import matplotlib.pyplot as plt
import numpy as np

# Crear matriz de ejemplo (10x10 con valores aleatorios)
data = np.random.rand(10, 10)
# Lista de algunos colormaps más usados en matplotlib
cmaps = [
"gray", "binary", "viridis", "plasma", "inferno", "magma", "cividis",
"Blues", "Greens", "Reds", "cool", "spring", "summer", "autumn", "winter",
"hot", "coolwarm", "bwr", "seismic", "tab10", "Set1", "Pastel1"
]
# Calcular el número de filas y columnas para mostrar en la cuadrícula
n_cols = 5
n_rows = int(np.ceil(len(cmaps) / n_cols))
# Crear subplots
fig, axes = plt.subplots(n_rows, n_cols, figsize=(15, 3*n_rows))
# Aplanar e iterar sobre los ejes
axes = axes.ravel()
for i, cmap in enumerate(cmaps):
	axes[i].imshow(data, cmap=cmap)
	axes[i].set_title(cmap, fontsize=10)
	axes[i].axis("off")
# Quitar ejes vacíos si sobran
for j in range(i+1, len(axes)):
	axes[j].axis("off")
plt.tight_layout()
plt.show()
```
