```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

# 1. Cargar imagen
img = cv2.imread('imagenes/6435ac1dc6711.jpg')

# Escala de grises
gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
plt.imshow(gray_img, cmap='gray')
plt.axis('off')
plt.show()

# 2. Blanco y Negro
_, binary_img = cv2.threshold(gray_img, 128, 255, cv2.THRESH_BINARY)
fig, axes = plt.subplots(2, 1, figsize=(15, 5))
axes[0].imshow(gray_img, cmap='gray')
axes[0].set_title("Escala de Grises")
axes[0].axis('off')
axes[1].imshow(binary_img, cmap='gray')
axes[1].set_title("Blanco y Negro")
axes[1].axis('off')
plt.show()

# 3. Reducción o Amplificación
resized_img = cv2.resize(img, (400, 400))
plt.imshow(cv2.cvtColor(resized_img, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()

# 4. Rotación
(h, w) = img.shape[:2]
center = (w // 2, h // 2)
M = cv2.getRotationMatrix2D(center, 45, 1.0)
rotated_img = cv2.warpAffine(img, M, (w, h))
plt.imshow(cv2.cvtColor(rotated_img, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()

# 5. Operaciones Aritméticas
img1 = cv2.imread('imagenes/6435ac1dc6711.jpg')
img2 = cv2.imread('imagenes/semaforo_1.jpg')
if img1 is None or img2 is None:
    print("❌ Una o ambas imágenes no se cargaron correctamente")

target_size = (400, 400)
img1_resized = cv2.resize(img1, target_size)
img2_resized = cv2.resize(img2, target_size)

sum_img = cv2.add(img1_resized, img2_resized)
diff_img = cv2.subtract(img1_resized, img2_resized)
mult_img = cv2.multiply(img1_resized, img2_resized)
div_img = cv2.divide(img1_resized, img2_resized)

fig, axes = plt.subplots(2, 2, figsize=(15, 8))
axes[0][0].imshow(cv2.cvtColor(sum_img, cv2.COLOR_BGR2RGB))
axes[0][0].set_title("Suma")
axes[0][0].axis('off')
axes[0][1].imshow(cv2.cvtColor(diff_img, cv2.COLOR_BGR2RGB))
axes[0][1].set_title("Resta")
axes[0][1].axis('off')
axes[1][0].imshow(cv2.cvtColor(mult_img, cv2.COLOR_BGR2RGB))
axes[1][0].set_title("Multiplicación")
axes[1][0].axis('off')
axes[1][1].imshow(cv2.cvtColor(div_img, cv2.COLOR_BGR2RGB))
axes[1][1].set_title("División")
axes[1][1].axis('off')
plt.tight_layout()
plt.show()

# 6. Operaciones Lógicas
img1 = cv2.imread('imagenes/6435ac1dc6711.jpg', cv2.IMREAD_GRAYSCALE)
img2 = cv2.imread('imagenes/semaforo_1.jpg', cv2.IMREAD_GRAYSCALE)
img1 = cv2.resize(img1, target_size)
img2 = cv2.resize(img2, target_size)

_, img1_bin = cv2.threshold(img1, 128, 255, cv2.THRESH_BINARY)
_, img2_bin = cv2.threshold(img2, 128, 255, cv2.THRESH_BINARY)

and_img = cv2.bitwise_and(img1_bin, img2_bin)
or_img = cv2.bitwise_or(img1_bin, img2_bin)
not_img1 = cv2.bitwise_not(img1_bin)
not_img2 = cv2.bitwise_not(img2_bin)

fig, axes = plt.subplots(2, 2, figsize=(15, 8))
axes[0][0].imshow(and_img, cmap='gray')
axes[0][0].set_title("AND")
axes[0][0].axis('off')
axes[0][1].imshow(or_img, cmap='gray')
axes[0][1].set_title("OR")
axes[0][1].axis('off')
axes[1][0].imshow(not_img1, cmap='gray')
axes[1][0].set_title("NOT IMAGEN 1")
axes[1][0].axis('off')
axes[1][1].imshow(not_img2, cmap='gray')
axes[1][1].set_title("NOT IMAGEN 2")
axes[1][1].axis('off')
plt.tight_layout()
plt.show()

# 7. Interpolación
resized_bilinear = cv2.resize(img, (400, 400), interpolation=cv2.INTER_LINEAR)
resized_nearest = cv2.resize(img, (400, 400), interpolation=cv2.INTER_NEAREST)

fig, axes = plt.subplots(2, 1, figsize=(15, 8))
axes[0].imshow(cv2.cvtColor(resized_bilinear, cv2.COLOR_BGR2RGB))
axes[0].set_title("Interpolación Bilineal")
axes[0].axis('off')
axes[1].imshow(cv2.cvtColor(resized_nearest, cv2.COLOR_BGR2RGB))
axes[1].set_title("Interpolación Vecino más cercano")
axes[1].axis('off')
plt.tight_layout()
plt.show()
```
